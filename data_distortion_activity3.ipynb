```python
#  All Tables: Run this code cell that makes pre-existing tools available for data exploration

import pandas as pd
import matplotlib.pyplot as plt
```

The DataFrame has many rows and 5 columns. Each row represents one month of unemployment data.

```python
# All Tables: Run this code cell that creates a pandas dataframe

df = pd.read_csv(
    'https://raw.githubusercontent.com/allegheny-college-cmpsc-105-spring-2026/activity03_dataDistortion_starter/main/data/data-distortion-bls-LNS14000000.csv',
    sep=',',
    names = ['Timepoint', 'Year', 'Month', 'Label', 'Value'],
    skiprows=1
)

print(df.head()) # Display the first few rows of the DataFrame
```

```python
-  In the variable explorer (`{x}`), what Shape does the variable `df` have? Write your response here:
```

## Table 1

```python
#  Run this code cell that creates a line graph of the data

# Assuming df is already created in the previous code cell
plt.close('all')

plt.figure(figsize=(12, 4))
plt.plot(df['Timepoint'], df['Value'], '*-')
plt.xticks(df['Timepoint'][::12], df['Label'][::12], rotation=0)
plt.xlabel('Timepoint (Years)')
plt.ylabel('Unemployment Rate (%)')
plt.title('Unemployment Rate Year to Year')
plt.grid(True)
plt.xlim(20,40)
plt.ylim(3,8)
plt.show()
```

-  What does this figure show?

This figure shows the unemployment rate over time with a focused range. The changes in unemployment are easy to see.
