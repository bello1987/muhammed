```python
#this code loads a CVS file from Github into a panda dataframe
"display the first row in dataframe"

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    'https://raw.githubusercontent.com/allegheny-college-cmpsc-105-spring-2026/dataDisplay-starter/main/data/data-display.csv',
    sep=',',
    names=['Mystery1', 'Mystery2', 'Mystery3', 'Mystery4'],
    skiprows=0
)

print(df.head()) # Display the first few rows of the DataFrame
```

```python
# A scatter plot of Mystery1 vs Mystery2  df
plt.figure()
plt.scatter(df['Mystery1'], df['Mystery2'])
plt.xlabel('Mystery1')
plt.ylabel('Mystery2')
plt.title('Scatter plot of mystery1 vs Mystery2')
plt.grid(True)
plt.show()
```

```python

plt.figure()
Counts = df['Mystery2'].value_counts()
plt.pie(Counts, labels=Counts.index)
plt.title('Distribution of Mystery2 values')
plt.show()
```

```python
# Assuming df is already created in the previous code
plt.figure()
plt.hist(df['Mystery4'], bins=20)
plt.xlabel('Mystery4')
plt.ylabel("Frequency")
plt.title(' Counts of Mystery4 Values')
plt.grid(True)
plt.show()
```

```python
# Assuming df is already created in the previous code
plt.figure()
plt.plot(df['Mystery2'], df['Mystery3'], marker='o', linestyle='-')
plt.xlabel('Mystery2')
plt.ylabel('Mystery3')
plt.title('Mystery3 as a function of Mytery3')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.show()
```

```python
# Assuming df is already created in the previous code
plt.figure()
plt.boxplot(df['Mystery3'])  # Create the boxplot #  Choose a column to visualize
plt.xlabel('Mystery3') #  Label the x-axis
plt.title('box plot of Mystery3 Values') #  Add a descriptive title
plt.show()
```
